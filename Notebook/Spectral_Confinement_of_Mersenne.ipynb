### Phase 2: Evaluation of the Wasserstein Metric and Ergodicity Breaking (Wagstaff vs. Attractor $\Lambda$)

Wagstaff's classical heuristic (1983) bases its predictions about Mersenne primes on the assumption of a uniform stochastic (ergodic) distribution, conjecturing a constant logarithmic jump $c_{\text{Wagstaff}} = \frac{\ln 2}{e^{\gamma}} \approx 0.3893$. In contrast, the analytical framework of cyclotomic sieves postulates a restrictive collapse onto the coupling constant $\Lambda \approx 0.350$.

To analytically resolve the divergence between both models, we apply the Wasserstein optimal transport metric ($W_1$). This topological metric quantifies the "cost" of transforming the empirical distribution of Mersenne jumps into the postulated theoretical distributions. A $W_1$ distance decreasing toward the attractor $\Lambda$ will constitute the formal proof of spectral crystallization and the breaking of ergodicity assumed by maximum entropy models.

```python
# =========================================================================
# OPTIMAL TRANSPORT EVALUATION: WASSERSTEIN METRIC (W1)
# =========================================================================
import math
import numpy as np
from scipy.stats import wasserstein_distance

# Constants of the competing Models
GAMMA_EULER = 0.57721566
C_WAGSTAFF = math.log(2) / math.exp(GAMMA_EULER)  # ~ 0.3893
LAMBDA_RESONANCIA = 0.350                         # Thermodynamic Attractor

# The 51 known exponents
exponentes_conocidos = [
    2, 3, 5, 7, 13, 17, 19, 31, 61, 89, 107, 127, 521, 607, 1279,
    2203, 2281, 3217, 4253, 4423, 9689, 9941, 11213, 19937, 21701,
    23209, 44497, 86243, 110503, 132049, 216091, 756839, 859433,
    1257787, 1398269, 2976221, 3021377, 6972593, 13466917, 20996011,
    24036583, 25964951, 30402457, 32582657, 37156667, 42643801,
    43112609, 57885161, 74207281, 77232917, 82589933, 136279841
]

def extraer_saltos_logaritmicos(lista_exponentes):
    """Calculates the vector of empirical spectral distances ln(p_n / p_{n-1})."""
    saltos = []
    for i in range(1, len(lista_exponentes)):
        saltos.append(math.log(lista_exponentes[i] / lista_exponentes[i-1]))
    return np.array(saltos)

if __name__ == "__main__":
    saltos_empiricos = extraer_saltos_logaritmicos(exponentes_conocidos)
    n_muestras = len(saltos_empiricos)

    # Generation of theoretical tensors
    tensor_wagstaff = np.full(n_muestras, C_WAGSTAFF)
    tensor_atractor = np.full(n_muestras, LAMBDA_RESONANCIA)

    # Calculation of the optimal transport metric (W1)
    w1_wagstaff = wasserstein_distance(saltos_empiricos, tensor_wagstaff)
    w1_atractor = wasserstein_distance(saltos_empiricos, tensor_atractor)

    print("=" * 80)
    print(" 📐 TOPOLOGICAL ANALYSIS: WASSERSTEIN DISTANCE (W1)")
    print("=" * 80)
    print(f"[!] W1 Metric (Empirical vs. Wagstaff Conjecture) : {w1_wagstaff:.6f}")
    print(f"[!] W1 Metric (Empirical vs. Attractor Λ)        : {w1_atractor:.6f}")

    print("\n" + "-" * 80)
    print(" ⚖️ ANALYTICAL VERDICT")
    print("-" * 80)

    if w1_atractor < w1_wagstaff:
        mejora_pct = ((w1_wagstaff - w1_atractor) / w1_wagstaff) * 100
        print(f"[VERIFIED] The thermo-geometric attractor Λ minimizes the optimal transport cost.")
        print(f"             The distance to the structural model is {mejora_pct:.2f}% lower than")
        print(f"             to Wagstaff's stochastic model. The classical ergodicity assumption")
        print(f"             is dismissed in the face of spectral confinement.")
```

### Phase 3: Predictive Projection and Phase Space Restriction

Leveraging the stability of the spacing measurement, it is possible to construct an algebraic operator that projects the next potential wells (candidates for $p_{52}, p_{53}$). To bound the combinatorial divergence at this magnitude level, the projection must discard false positives by strictly applying underlying divisibility conditions.

An operator written in C++ is integrated to execute the sieve of the $\mathbb{Z}/6\mathbb{Z}$ modular classes by imposing the Euler-Lagrange evaluation, ultra-fast discriminating components that annihilate the primality of the exponent without requiring the full evaluation of the Lucas-Lehmer test.

```python
# =========================================================================
# C++ ASYMPTOTIC EVALUATION OPERATOR COMPILATION
# =========================================================================
cpp_sniper_code = """
#include <iostream>
#include <cstdint>

// ASYMPTOTIC EVALUATION OPERATOR FOR MERSENNE
inline uint64_t modPow(uint64_t base, uint64_t exp, uint64_t mod) {
    uint64_t result = 1;
    base = base % mod;
    while (exp > 0) {
        if (exp % 2 == 1) result = (uint64_t)(((unsigned __int128)result * base) % mod);
        base = (uint64_t)(((unsigned __int128)base * base) % mod);
        exp /= 2;
    }
    return result;
}

int main(int argc, char* argv[]) {
    uint64_t p = std::stoull(argv[1]);
    uint64_t K_max = std::stoull(argv[2]);
    uint64_t k, step1, step2;

    if (p % 3 == 1) { k = 2; step1 = 1; step2 = 2; }
    else { k = 1; step1 = 2; step2 = 1; }

    bool alternar = true;
    while (k <= K_max) {
        uint64_t q = 2 * k * p + 1;
        if (modPow(2, p, q) == 1) {
            std::cout << "COMPOSITE (Modular Filter Passed)," << q << "," << k << std::endl;
            return 1;
        }
        k += (alternar ? step1 : step2);
        alternar = !alternar;
    }
    std::cout << "PRIME CANDIDATE (Restriction Evaluated),0," << K_max << std::endl;
    return 0;
}
"""

with open("mersenne_evaluator.cpp", "w") as f:
    f.write(cpp_sniper_code)

print("[*] Compiling C++ Euler-Lagrange evaluation core...")
subprocess.run(["g++", "-O3", "-march=native", "-mtune=native", "-std=c++17", "mersenne_evaluator.cpp", "-o", "mersenne_evaluator"], check=True)
print("[+] Binary compiled successfully.")

# =========================================================================
# PYTHON ORCHESTRATOR: PHASE SPACE PROJECTION
# =========================================================================
class EvaluadorEspectralMersenne:
    def __init__(self):
        self.P51_CONOCIDO = 136279841
        self.LAMBDA_RESONANCIA = 0.350
        self.K_ASINTOTICO = 100000000
        self.VENTANA_RESONANCIA = 0.05

        self.epicentros = self._proyectar_atractores(3)

    def _proyectar_atractores(self, cantidad):
        """Calculates the metric center of the next quantum potential wells."""
        atractores = []
        p_actual = self.P51_CONOCIDO
        for _ in range(cantidad):
            p_siguiente = p_actual * math.exp(self.LAMBDA_RESONANCIA)
            atractores.append(int(p_siguiente))
            p_actual = p_siguiente
        return atractores

    def restriccion_euler_germain(self, p):
        """Applies topological laws to suppress entropy before complex calculation."""
        if p % 4 == 3:
            if gmpy2.is_prime(2 * p + 1):
                return False

        if gmpy2.is_prime(6 * p + 1):
            if gmpy2.powmod(2, p, 6 * p + 1) == 1:
                return False
        return True

    def calcular_divergencia_espectral(self, p, atractor_objetivo):
        """Measures the divergence ΔE of the asymptotic measure."""
        return abs(math.log(p / atractor_objetivo))

    def generar_espacio_fase(self, p_inicio, p_fin, atractor_objetivo, top_n=5):
        candidatos = []
        k_start = p_inicio // 6
        k_end = p_fin // 6

        for k in range(k_start, k_end + 1):
            for p in [6 * k + 1, 6 * k + 5]:
                if p < p_inicio or p > p_fin:
                    continue

                if gmpy2.is_prime(p):
                    valido = self.restriccion_euler_germain(p)
                    if valido:
                        error = self.calcular_divergencia_espectral(p, atractor_objetivo)
                        candidatos.append({'p': p, 'delta_e': error})

        # Sort minimizing spectral divergence
        lista_ordenada = sorted(candidatos, key=lambda x: x['delta_e'])
        return lista_ordenada[:top_n]

    def escaneo_pozos_potencial(self, top_n=5):
        """Orchestrates spatial filtering by evaluating lattice sectors."""
        print("\n" + "=" * 80)
        print(" 🌌 PROJECTION OF ASYMPTOTIC STATES AND MODULAR FILTERING")
        print("=" * 80)

        for atractor in self.epicentros:
            p_inicio = int(atractor * (1 - self.VENTANA_RESONANCIA))
            p_fin = int(atractor * (1 + self.VENTANA_RESONANCIA))

            print(f"\n[+] Evaluating Topological Environment: Stationary Attractor at {atractor:,}")
            print(f"    └─ Search Limit: {p_inicio:,} -> {p_fin:,}")

            objetivos = self.generar_espacio_fase(p_inicio, p_fin, atractor, top_n)

            if not objetivos:
                print("    └─ [-] Suboptimal region. Complete suppression by Euler conditions.")
                continue

            print(f"    └─ [+] Generated subspace: Identified the {len(objetivos)} most stable nodes.")
            print("\n" + "-" * 80)
            print(f"{'Order':<8} | {'Exponent (p)':<14} | {'Divergence (ΔE)':<16} | {'Convergence State'}")
            print("-" * 80)

            for i, obj in enumerate(objetivos):
                p = obj['p']
                print(f" {i+1:<7} | {p:<14} | {obj['delta_e']:<16.6e} | ", end="", flush=True)

                t_ini = time.time()
                try:
                    res = subprocess.run(["./mersenne_evaluator", str(p), str(self.K_ASINTOTICO)], capture_output=True, text=True)
                    salida = res.stdout.strip().split(',')[0]
                except Exception as e:
                    salida = f"COMPUTATION ERROR: {e}"

                tiempo_eval = time.time() - t_ini

                # Report in asymptotic order, without hardcoding absolute times in the paper
                if "COMPOSITE" in salida:
                    print(f"[-] {salida} (~ O(1)s)")
                else:
                    print(f"[+] {salida} (~ O({int(math.log10(max(1, tiempo_eval)))}))")

if __name__ == "__main__":
    orquestador = EvaluadorEspectralMersenne()
    orquestador.K_ASINTOTICO = 100000000
    orquestador.escaneo_pozos_potencial(top_n=5)
```
