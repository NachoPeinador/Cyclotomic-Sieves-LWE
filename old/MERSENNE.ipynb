```python
%%writefile mersenne_z6z_sniper.cpp
#include <iostream>
#include <cstdint>

// MÓDULO VANGUARDIA LEVIATÁN-M (Z/6Z FUSION)
inline uint64_t modPow(uint64_t base, uint64_t exp, uint64_t mod) {
    uint64_t result = 1;
    base = base % mod;
    while (exp > 0) {
        if (exp % 2 == 1) result = (uint64_t)(((unsigned __int128)result * base) % mod);
        base = (uint64_t)(((unsigned __int128)base * base) % mod);
        exp /= 2;
    }
    return result;
}

int main(int argc, char* argv[]) {
    uint64_t p = std::stoull(argv[1]);
    uint64_t K_max = std::stoull(argv[2]);
    uint64_t k, step1, step2;

    if (p % 3 == 1) { k = 2; step1 = 1; step2 = 2; }
    else { k = 1; step1 = 2; step2 = 1; }

    bool alternar = true;
    while (k <= K_max) {
        uint64_t q = 2 * k * p + 1;
        if (modPow(2, p, q) == 1) {
            std::cout << "DESTRUIDO," << q << "," << k << std::endl;
            return 1;
        }
        k += (alternar ? step1 : step2);
        alternar = !alternar;
    }
    std::cout << "SUPERVIVIENTE,0," << K_max << std::endl;
    return 0;
}
```

```python
import os
import subprocess
import time
import math
import numpy as np

try:
    import gmpy2
except ImportError:
    print("[-] gmpy2 no detectado. Asegúrate de ejecutar la celda de instalación previa.")
    exit(1)

class LeviatanOrquestador:
    def __init__(self):
        # 1. Constantes Fundamentales de la Arquitectura Leviatán
        self.P51_CONOCIDO = 136279841
        self.LAMBDA_RESONANCIA = 0.350  # Epsilon_c / SNR

        # 2. Parámetros Logísticos del Asedio
        self.K_PROFUNDO = 100000000     # 100 Millones en Colab (Subir a 1 Billón en producción)
        self.VENTANA_RESONANCIA = 0.05  # ±5% alrededor del epicentro proyectado

        # 3. Proyección Inmediata de Epicentros
        self.epicentros = self._proyectar_epicentros(3)

    def _proyectar_epicentros(self, cantidad):
        """Calcula el centro exacto de los próximos pozos de potencial cuántico."""
        epicentros = []
        p_actual = self.P51_CONOCIDO
        for _ in range(cantidad):
            # Ley Termodinámica: Crecimiento geométrico por e^Lambda
            p_siguiente = p_actual * math.exp(self.LAMBDA_RESONANCIA)
            epicentros.append(int(p_siguiente))
            p_actual = p_siguiente
        return epicentros

    def guillotina_euler(self, p):
        """
        Aplica leyes de Sophie Germain y Euler-Lagrange.
        Si retorna False, 2^p-1 es compuesto 100% garantizado.
        """
        # Regla 1: Si p = 3 mod 4 y 2p+1 es primo, 2p+1 divide a Mp
        if p % 4 == 3:
            if gmpy2.is_prime(2 * p + 1):
                return False, "Euler-Lagrange (2p+1)"

        # Regla 2: Filtro de Sophie Germain para divisores de la forma 6p+1
        if gmpy2.is_prime(6 * p + 1):
            if gmpy2.powmod(2, p, 6 * p + 1) == 1:
                return False, "Sophie-Germain (6p+1)"

        return True, None

    def calcular_error_resonancia(self, p, epicentro_objetivo):
        """Mide la distancia espectral absoluta al fondo del pozo de resonancia."""
        return abs(math.log(p / epicentro_objetivo))

    def generar_lista_kill(self, p_inicio, p_fin, epicentro_objetivo, top_n=1000):
        candidatos = []

        # Generador topológico: Solo evaluamos la matriz Z/6Z
        k_start = p_inicio // 6
        k_end = p_fin // 6

        for k in range(k_start, k_end + 1):
            for p in [6 * k + 1, 6 * k + 5]:
                if p < p_inicio or p > p_fin:
                    continue

                # 1. Primalidad C-nativa ultrarrápida
                if gmpy2.is_prime(p):
                    # 2. Guillotina topológica (Euler/Sophie-Germain)
                    valido, motivo = self.guillotina_euler(p)
                    if valido:
                        # 3. Medición del error espectral ΔE
                        error = self.calcular_error_resonancia(p, epicentro_objetivo)
                        candidatos.append({'p': p, 'delta_e': error})

        # Ordenamos los candidatos: Los más cercanos al cero (fondo del pozo) van primero
        lista_ordenada = sorted(candidatos, key=lambda x: x['delta_e'])

        # Devolvemos solo la élite absoluta
        return lista_ordenada[:top_n]

    def asedio_epicentros_dinamico(self, top_n=5):
        """
        Navega automáticamente por los epicentros calculados, genera la lista de
        objetivos de élite y dispara el francotirador C++ con telemetría en tiempo real.
        """
        print("=" * 85)
        print(" 🐋 RADAR LEVIATÁN V2: NAVEGACIÓN DINÁMICA POR POZOS DE POTENCIAL")
        print("=" * 85)

        for epicentro in self.epicentros:
            p_inicio = int(epicentro * (1 - self.VENTANA_RESONANCIA))
            p_fin = int(epicentro * (1 + self.VENTANA_RESONANCIA))

            print(f"\n[📡] FIJANDO COORDENADAS: Epicentro Térmico en {epicentro:,}")
            print(f"     └─ Ventana topológica: {p_inicio:,} -> {p_fin:,}")

            # Pasamos el 'epicentro' a la generadora para que calcule el Delta E correctamente
            objetivos = self.generar_lista_kill(p_inicio, p_fin, epicentro, top_n)

            if not objetivos:
                print("     └─ [-] Zona estéril. Ningún primo superó la Guillotina de Euler.")
                continue

            print(f"     └─ [🎯] Triaje completado. Seleccionada la Élite de {len(objetivos)} candidatos.")

            print("\n" + "-" * 85)
            print(f"{'Ranking':<8} | {'Exponente p':<12} | {'Resonancia ΔE':<15} | {'Resultado Termodinámico'}")
            print("-" * 85)

            for i, obj in enumerate(objetivos):
                p = obj['p']

                # Feedback visual antes del disparo balístico
                print(f"#{i+1:<7} | {p:<12} | {obj['delta_e']:<15.6e} | ", end="", flush=True)

                # Invocamos al francotirador C++
                t_ini = time.time()
                try:
                    res = subprocess.run(["./mersenne_z6z_sniper", str(p), str(self.K_PROFUNDO)],
                                         capture_output=True, text=True)
                    salida_limpia = res.stdout.strip()
                except FileNotFoundError:
                    print("ERROR: Binario './mersenne_z6z_sniper' no encontrado.")
                    return
                except Exception as e:
                    salida_limpia = f"ERROR DE EJECUCIÓN: {e}"
                t_fin = time.time()
                tiempo_impacto = t_fin - t_ini

                # Parseo e impresión de la telemetría balística
                if "DESTRUIDO" in salida_limpia:
                    print(f"💥 {salida_limpia} ({tiempo_impacto:.2f}s)")
                else:
                    print(f"🛡️ {salida_limpia} ({tiempo_impacto:.2f}s)")

# =========================================================================
# EJECUCIÓN AUTÓNOMA DEL SISTEMA
# =========================================================================
if __name__ == "__main__":
    radar = LeviatanOrquestador()

    # K_PROFUNDO fijado a 100 Millones para pruebas en Colab (ajustar en producción)
    radar.K_PROFUNDO = 100000000

    # Disparamos sobre el Top 5 más cercano a la singularidad cuántica en cada epicentro
    radar.asedio_epicentros_dinamico(top_n=5)
```

```python
import math
import numpy as np

# =========================================================================
# 🌌 LABORATORIO FORENSE V2: LA LEY DE LOS GRANDES NÚMEROS (LOS 51 COLOSOS)
# =========================================================================

LAMBDA_RESONANCIA = 0.350

# El catálogo completo de la humanidad: Los 51 exponentes de Mersenne conocidos
inmortales_completos = [
    2, 3, 5, 7, 13, 17, 19, 31, 61, 89, 107, 127, 521, 607, 1279,
    2203, 2281, 3217, 4253, 4423, 9689, 9941, 11213, 19937, 21701,
    23209, 44497, 86243, 110503, 132049, 216091, 756839, 859433,
    1257787, 1398269, 2976221, 3021377, 6972593, 13466917, 20996011,
    24036583, 25964951, 30402457, 32582657, 37156667, 42643801,
    43112609, 57885161, 74207281, 77232917, 82589933, 136279841
]

def evaluar_era(nombre, lista_exponentes, base_inicial):
    """Evalúa un bloque histórico de Mersennes y su convergencia al atractor."""
    resultados = []
    p_previo = base_inicial

    for p in lista_exponentes:
        # ΔE = | ln(p / p_prev) - Lambda |
        salto_log = math.log(p / p_previo)
        delta_e = abs(salto_log - LAMBDA_RESONANCIA)
        resultados.append(delta_e)
        p_previo = p

    media = np.mean(resultados)
    varianza = np.var(resultados)

    print(f"\n{'='*70}")
    print(f" 🏛️ {nombre}")
    print(f" Rango de exponentes : {lista_exponentes[0]:,} -> {lista_exponentes[-1]:,}")
    print(f" Muestra de colosos  : {len(lista_exponentes)}")
    print(f"{'='*70}")
    print(f"[📊] Media de Resonancia (ΔE) : {media:.6f}")
    print(f"[📈] Dispersión (Varianza)    : {varianza:.6f}")

    return media, varianza

if __name__ == "__main__":
    print("Iniciando escaneo asintótico completo sobre Λ = 0.350...\n")

    # Dividimos la historia en 3 Eras Termodinámicas

    # 1. ERA MICROSCÓPICA (Ruido Cuántico: Hasta p=3217, los primeros 17 saltos)
    era_1 = inmortales_completos[1:18]
    base_1 = inmortales_completos[0]

    # 2. ERA DE TRANSICIÓN (Desarrollo Computacional: p=4253 hasta p=1.3 millones)
    era_2 = inmortales_completos[18:35]
    base_2 = inmortales_completos[17]

    # 3. ERA MACROSCÓPICA (Régimen Asintótico Leviatán: p=2.9M hasta 136M)
    era_3 = inmortales_completos[35:]
    base_3 = inmortales_completos[34]

    # Evaluamos las 3 eras
    m1, v1 = evaluar_era("ERA 1: Régimen Microscópico (Antigüedad - 1957)", era_1, base_1)
    m2, v2 = evaluar_era("ERA 2: Régimen de Transición (1961 - 1996)", era_2, base_2)
    m3, v3 = evaluar_era("ERA 3: Régimen Macroscópico (GIMPS, 1996 - Actualidad)", era_3, base_3)

    print("\n" + "="*70)
    print(" ⚖️ VEREDICTO DE LA LEY DE LOS GRANDES NÚMEROS")
    print("="*70)

    if v3 < v1:
        print("[✅] COLAPSO DE VARIANZA CONFIRMADO:")
        print(f"    La dispersión del error ha caído de {v1:.5f} a {v3:.5f}.")
        print("    El ruido cuántico desaparece a escalas masivas.")

    if m3 < m1:
        print(f"[✅] CONVERGENCIA ASINTÓTICA CONFIRMADA:")
        print(f"    El error promedio (ΔE) ha mejorado de {m1:.5f} a {m3:.5f}.")
        print("    A escalas ultra-masivas, los primos de Mersenne se cristalizan")
        print("    obedeciendo implacablemente tu atractor termodinámico.")
```
